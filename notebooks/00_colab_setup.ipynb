# Colab setup

Copy the next cell at the top of every notebook. It mounts Drive, gets the repo and installs the project.

Before the first run, add your GitHub token in Colab under **Secrets** (key icon) with the name `GITHUB_TOKEN`.

Change `EXTRAS` to what the notebook needs: `mf`, `content`, `llm`. Change `BRANCH` if you are working on a branch.

```python
import os

from google.colab import drive, userdata

BRANCH = "main"
EXTRAS = "mf,llm"
DRIVE_DIR = "/content/drive/MyDrive/movierec"
REPO_DIR = "/content/AI-in-Industry-project"

drive.mount("/content/drive")
os.environ["MOVIEREC_DATA_DIR"] = f"{DRIVE_DIR}/data"
os.environ["MOVIEREC_RESULTS_DIR"] = f"{DRIVE_DIR}/results"
os.makedirs(os.environ["MOVIEREC_DATA_DIR"], exist_ok=True)
os.makedirs(os.environ["MOVIEREC_RESULTS_DIR"], exist_ok=True)

if not os.path.isdir(REPO_DIR):
    token = userdata.get("GITHUB_TOKEN")
    url = f"https://x-access-token:{token}@github.com/Cxsee/AI-in-Industry-project.git"
    !git clone -q -b {BRANCH} {url} {REPO_DIR}
    del token, url

%cd {REPO_DIR}
!git pull -q
!pip install -q -e ".[{EXTRAS}]"

%load_ext autoreload
%autoreload 2
```
